```python
from datasets import load_dataset
from pathlib import Path
import pandas as pd

# Load dataset
ds = load_dataset("Dasool/VERI-Emergency")

# Choose split
split = "train"

# Create image output folder
image_dir = Path("VERI_images")
image_dir.mkdir(parents=True, exist_ok=True)

rows = []

for i, example in enumerate(ds[split]):
    image = example["image"]
    image_id = example["image_id"]

    # Make a safe filename
    safe_image_id = (
        str(image_id)
        .replace("/", "_")
        .replace("\\", "_")
        .replace(" ", "_")
    )

    image_path = image_dir / f"{safe_image_id}.jpg"

    # Save image to VERI_images/
    image.save(image_path)

    rows.append({
        "image_path": str(image_path),
        "image_id": image_id,
        "risk_identification": example["risk_identification"],
        "category": example["category"],
        "caption": example["caption"],
        "emergency_response": example["emergency_response"],
    })

# Save metadata CSV outside the image folder
metadata = pd.DataFrame(rows)
metadata.to_csv("VERI_metadata.csv", index=False, encoding="utf-8")

print(f"Saved {len(rows)} images to {image_dir}")
print("Saved metadata to VERI_metadata.csv")
print(metadata.head())
```
